```python
import anthropic
client = anthropic.Anthropic()

def ask(prompt, model="claude-sonnet-5-5", max_tokens=1000):
    msg = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        messages=[{"role": "user", "content": prompt}],
    )
    return msg.content[0].text
```

```python
print(ask("pandas로 csv 파일의 결측치 확인하는 코드를 알려줘"))
```
