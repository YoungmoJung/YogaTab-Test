```python
import pandas as pd
from graph_store import GraphStore

g = GraphStore()
with g.driver.session() as s:
    nodes = pd.DataFrame(s.run(
        "Match (n) RETURN labels(n)[0] AS label, n.name AS name, properties(n) AS props" 
    ).data())
    rels = pd.DataFrame(s.run(
        "MATCH (a)-[r]->(b) RETURN a.name AS 시작, type(r) AS 관계, b.name as 끝"
    ).data())
g.close()

display(nodes)
display(rels)
```
